## 1. User

`User` contains the data and behaviour shared by customers and delivery partners.

Keep `User` abstract. `add_to_wallet(amount)` must accept zero or positive values. A negative amount must not change the wallet balance.

### Predict the Output

Before running the statement below, write down what you expect to happen.

```python
u = User("Aman", "9999999999")
```

````python
from abc import ABC, abstractmethod

class User(ABC):

    def __init__(self, name, phone):
        self._name = name
        self._phone = phone
        self._wallet_balance = 0

    def add_to_wallet(self, amount):
        if amount >= 0:
          self._wallet_balance += amount

    @abstractmethod
    def notify(self, message):
        pass

    @abstractmethod
    def display_profile(self):
        pass
````

## 2. Customer

`Customer` inherits from `User`.

Additional data:
- `address`
- `order_history`, initially an empty list

Required behaviour:
- `notify(message)` prints a customer-specific notification.
- `display_profile()` prints the customer name, phone, wallet balance, and address.
- `place_order(restaurant, items)` creates an `Order`, adds it to `order_history`, and returns the created order.

````python
class Customer(User):

    def __init__(self, name, phone, address):
       super().__init__(name,phone)
       self.address = address
       self.order_history = []

    def notify(self, message):
        print("Notificatoin: ",message)

    def display_profile(self):
        print("Name: ", self._name)
        print("Phone: ", self._phone)
        print("Wallet Balance: ", self._wallet_balance)
        print("Address: ", self.address)

    def place_order(self, restaurant, items):
        order = Order(items)
        self.order_history.append(order)
        print("Your OTP is: ",order._otp)

        return order
````

````python
#@title Validate User and Customer { display-mode: "form" }

def _check_user_customer():
    messages = []

    try:
        User("Aman", "9999999999")
        messages.append("CHECK: User should not be created directly. Keep User abstract.")
    except TypeError:
        messages.append("PASS: User cannot be created directly.")
    except Exception:
        messages.append("CHECK: User creation is failing for a reason other than abstraction.")

    try:
        customer = Customer("Priya", "9876543210", "Bangalore")

        if customer._wallet_balance == 0:
            messages.append("PASS: Customer wallet starts at 0.")
        else:
            messages.append("CHECK: Customer wallet should start at 0.")

        customer.add_to_wallet(500)

        if customer._wallet_balance == 500:
            messages.append("PASS: Wallet top-up works.")
        else:
            messages.append("CHECK: add_to_wallet() is not updating the balance correctly.")

        before = customer._wallet_balance

        try:
            customer.add_to_wallet(-100)
        except Exception:
            pass

        if customer._wallet_balance == before:
            messages.append("PASS: Negative wallet top-up is rejected.")
        else:
            messages.append("CHECK: A negative top-up must not change the wallet balance.")

    except Exception:
        messages.append("CHECK: Customer could not be created. Review the constructor and User inheritance.")

    print("\n".join(messages))

_check_user_customer()
````

````python
class Restaurant:

    def __init__(self, name, location):
        self.name = name
        self.location = location
        self._menu = []

    def add_item(self, menu_item):
        self._menu.append(menu_item)

    def get_menu(self):
        return self._menu

    def is_open(self):
        return True
````

### Order Requirements

**Attributes**

- `_order_id`, generated automatically
- `_items`, containing the selected `MenuItem` objects
- `_status`, initially `Placed`
- `_otp`, fixed at `1234`

**Methods**

- `calculate_bill()`
- `estimated_time()`
- `update_status(new_status)`
- `verify_otp(otp)`

**Billing rule**

Subtotal + 5% GST + ₹20 packaging fee.

`estimated_time()` returns `30`.

````python
# import random
# self._otp = random.randint(1000,9999)
````

````python

import random

class Order:

    _counter = 1

    def __init__(self, items):
        self._order_id = Order._counter
        Order._counter += 1
        self._items = items
        self._status = "Placed"
        self._otp = random.randint(1000,9999)



    def calculate_bill(self):
        subtotal = 0
        for item in self._items:
          subtotal += item.price

        gst = subtotal * 0.05
        packaging_fees = 20

        total = subtotal + gst + packaging_fees

        return total

    def estimated_time(self):
        return 30

    def update_status(self, new_status):
      if new_status in ['Placed','Order Accepted','Processing','Cancelled','Delivered']:
        self._status = new_status

    def verify_otp(self, otp):
      if otp == self._otp:
        return True
      else:
        return False
````

The accepted values are `Placed`, `Accepted`, and `Delivered`.

````python
# Test the order

item1 = MenuItem("Biryani", 250, False)
item2 = MenuItem("Kebab", 180, False)

order = Order([item1, item2])

print("Bill:", order.calculate_bill())
print("Estimated time:", order.estimated_time())
print("OTP valid:", order.verify_otp(1234))
````

````python
#@title Validate Order { display-mode: "form" }

def _check_order():
    messages = []

    try:
        item1 = MenuItem("Biryani", 250, False)
        item2 = MenuItem("Kebab", 180, False)
        order = Order([item1, item2])

        total = order.calculate_bill()

        if abs(total - 471.5) < 0.001:
            messages.append("PASS: Bill calculation is correct.")
        else:
            messages.append("CHECK: Order bill is off. Recheck subtotal, 5% GST, and the ₹20 packaging fee.")

        if order.estimated_time() == 30:
            messages.append("PASS: Estimated time is correct.")
        else:
            messages.append("CHECK: estimated_time() should return 30.")

        if order._status == "Placed":
            messages.append("PASS: New order starts as Placed.")
        else:
            messages.append("CHECK: A new order should start with status Placed.")

        if order.verify_otp(9999) is False:
            messages.append("PASS: Wrong OTP is rejected.")
        else:
            messages.append("CHECK: verify_otp() should return False for a wrong OTP.")

        if order.verify_otp(1234) is True:
            messages.append("PASS: Correct OTP is accepted.")
        else:
            messages.append("CHECK: verify_otp() should return True for OTP 1234.")

    except Exception:
        messages.append("CHECK: Order could not be tested. Review the constructor and required methods.")

    print("\n".join(messages))

_check_order()
````

## 6. DeliveryPartner

`DeliveryPartner` inherits from `User`.

Additional attributes:

- `vehicle`
- `is_available`, initially `True`
- `rating`, initially `0.0`

Required behaviour:

**`accept_order(order)`**
- changes the order status to `Accepted`
- makes the delivery partner unavailable

**`deliver(order, otp)`**
- verifies the OTP before completing delivery
- leaves the order as `Accepted` when the OTP is incorrect
- keeps the delivery partner unavailable when the OTP is incorrect
- changes the order status to `Delivered` when the OTP is correct
- makes the delivery partner available again after successful delivery

````python
class DeliveryPartner(User):

    def init(self, name, phone, vehicle):

        super().__init__(name,phone)

        self.vehicle = vehicle

        self.is_available = True

        self.rating = 0.0

    def notify(self, message):

        self.message = message

        print("Notification: ", message, "for delivery partner: ", self._name)

    def display_profile(self):

        print("Name: ", self._name)

        print("Phone: ", self._phone)

        print("availibility: ", self.is_available)

        print("Rating: ", self.rating)

    def accept_order(self, order):

        order.update_status("Accepted")

        self.is_available = False

    def deliver(self, order, otp):

      if order.verify_otp(otp):
        order.update_status("Delivered")
        self.is_available = True
````

````python
#@title Final validation { display-mode: "form" }

def _final_project_check():
    messages = []

    try:
        customer = Customer("Priya", "9876543210", "Bangalore")
        partner = DeliveryPartner("Rajesh", "9999999999", "Bike")

        item1 = MenuItem("Biryani", 250, False)
        item2 = MenuItem("Kebab", 180, False)

        restaurant = Restaurant("Bawarchi", "MG Road")
        restaurant.add_item(item1)
        restaurant.add_item(item2)

        customer.add_to_wallet(500)

        before = customer._wallet_balance
        try:
            customer.add_to_wallet(-100)
        except Exception:
            pass

        if customer._wallet_balance == before:
            messages.append("PASS: Negative wallet top-up is rejected.")
        else:
            messages.append("CHECK: A negative wallet top-up must not change the balance.")

        order = customer.place_order(restaurant, [item1, item2])

        if isinstance(order, Order):
            messages.append("PASS: place_order() returns an Order.")
        else:
            messages.append("CHECK: place_order() should return an Order object.")

        if order in customer.order_history:
            messages.append("PASS: Order is stored in customer order history.")
        else:
            messages.append("CHECK: The created order should be added to order_history.")

        if abs(order.calculate_bill() - 471.5) < 0.001:
            messages.append("PASS: Final bill is correct.")
        else:
            messages.append("CHECK: Final bill should be 471.5.")

        partner.accept_order(order)

        if order._status == "Accepted" and partner.is_available is False:
            messages.append("PASS: Delivery partner accepts the order correctly.")
        else:
            messages.append("CHECK: accept_order() should set the order to Accepted and make the partner unavailable.")

        partner.deliver(order, 9999)

        if order._status == "Accepted" and partner.is_available is False:
            messages.append("PASS: Wrong OTP does not complete delivery.")
        else:
            messages.append("CHECK: Wrong OTP should leave the order Accepted and the partner unavailable.")

        partner.deliver(order, 1234)

        if order._status == "Delivered" and partner.is_available is True:
            messages.append("PASS: Correct OTP completes delivery.")
        else:
            messages.append("CHECK: OTP 1234 should mark the order Delivered and make the partner available.")

        import io
        import contextlib

        customer_output = io.StringIO()
        partner_output = io.StringIO()

        with contextlib.redirect_stdout(customer_output):
            customer.notify("Order delivered")

        with contextlib.redirect_stdout(partner_output):
            partner.notify("Order delivered")

        if customer_output.getvalue().strip() != partner_output.getvalue().strip():
            messages.append("PASS: Customer and DeliveryPartner notifications are different.")
        else:
            messages.append("CHECK: Customer and DeliveryPartner notifications should be visibly different.")

    except Exception:
        messages.append("CHECK: The complete project could not be tested. Review the earlier validation results.")

    print("\n".join(messages))

_final_project_check()
````
